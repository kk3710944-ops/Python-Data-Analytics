```python
import numpy as np
```

```python
a=np.array([10,20,30,40,55])
#print first,third,last element
print(a[0:5:2])
#print the last 3 numbers
print(a[2:5])
#print reverse the array
print(a[::-1])
#add one element
a=np.append(a,60)
print(a)
#print the number even indexing
print(a[::2])
#print the even numbers
print(a[a%2==0])
#print the number >30
print(a[a>30])
#convbert the negative numbers into 0.
b=np.array([12,23,-3,-5,34])
b[b<0]=0
print(b)
#sum the array elements
print(sum(a))
#find mean,midian,maxinum and minimum value
print(np.mean(a))
print(np.median(a))
print(np.max(a))
print(np.min(a))
#find the index of max value using np.argmax()
print(np.argmax(a))
#find the index of min value using np.argmax()
print(np.argmin(a))
#calculate the standard deviation using np.std()
print(np.std(a))
ar=np.array([10,60,30,80,45,90])
print(ar)
#find all element >50.
print(ar[ar>50])
#find all even numbers
print(ar[ar%2==0])
#find values between 20 to 60
print(ar[(ar>=20)&(ar<=60)])
#create number from 1 to 12 and reshape them into 3,4 matrix
arr=np.arange(1,13).reshape(3,4)
print(arr)
print(sum(arr[0]))
print(sum(arr[1]))
print(sum(arr[2]))

```

```python
a=np.arange(11,20).reshape(3,3)
print(a)
#print(a[0][0],a[0,2],a[2][0],a[2][2])
print('matrix')
for i in range(0,3):
    print(a[i][i])
print('matrix-1')
for i in range(0,2):
    print(a[i][i]+1)
print('matrix-2')
for i in range(1,3):
    print(a[i][i]-1)
```

```python
a=np.array([[2,3],[4,5]])
print(a)
b=np.array([[5,6]])
print(b)
print(np.vstack((a,b)))
print(np.hstack((a,a)))
```
