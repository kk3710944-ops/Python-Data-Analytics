```python
x=np.arange(0,10)
print(x)
y=np.arange(11,21)
print(y)
```

```python
mp.scatter(x,y)
mp.xlabel('xeyy')
mp.ylabel('yeyy')
mp.title('graph in 2D')
```

```python
#data to plot
l=['python','java','c++','c']
sizes=[250,240,124,345]
#e=(2,0.20,0.4,0.7)
mp.pie(sizes,labels=l,autopct='%.0f%%')
#e
```

```python
#subplot used to create multiple plot in on cell.
mp.subplot(2,2,1)
mp.scatter(x,y)

mp.subplot(2,2,2)
mp.plot(x,y,'r*--')

mp.subplot(2,2,3)
mp.plot(x,y,'go')

mp.subplot(2,2,4)
mp.bar(x,y)
```
