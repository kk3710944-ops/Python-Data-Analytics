```python
#Convert Two Lists into a Dictionary
a=['name','age','depatrment']
b=['ram',20,'maths']
f={}
for i in range(len(a)):
    f[a[i]]=b[i]
print(f)
```

```python
#Find Duplicate Names and Their Count names = ["Asha", "Rahul", "Asha", "Priya", "Rahul", "Asha"] output-{'Asha': 3, 'Rahul': 2, 'Priya': 1} 
names = ["Asha", "Rahul", "Asha", "Priya", "Rahul", "Asha"] 
name1={}
for i in names:
    if i in name1:
            name1[i]+=1
    else:
            name1[i]=1
print(name1)

```

```python
#Find the Key with Highest Value
b={'a':1,'b':2,'c':4,'d':6}
max_key=''
max_value=0
for k,v in b.items():
    if v>max_value:
        max_value=v
        max_key=k
print(max_key)
print(max_value)
```

```python
#fetch the even number from a list using def.
a=[1, 2,3,4,6,8,9,10]
def even(a):
    for i in a:
        if i%2==0:
            print(i)
even(a)
```

```python
#Separate Even and Odd Numbers from a list create to separate list.
a=[12,13,14,43,50,10]
odd=[]
even=[]
for i in range(len(a)):
    if a[i]%2==0:
        even.append(a[i])
    else:
        odd.append(a[i])
print(even)
print(odd)
```

```python
#list common value of 2 list.
a=[12,13,14,43,50,10]
b=[12,40,10,5]
c=[]
for i in a:
    if i in b:
        c.append(i)

print(c)
```

```python
#remove duplicate from the list
a=[12,23,12,10,5,15,10,17]
unique=[]
for i in a:
    if i not in unique:
        unique.append(i)
print(unique)
```

```python
#swap first and last numbers using indexing
a=[12,23,54,60,70,99,37]
a[0],a[-1]=a[-1],a[0]
print(a)
```
