```python
df=se.load_dataset('tips')
df
df.describe()
df.duplicated().sum()
df.drop_duplicates()
df.info()
```

```python
mp.subplot(2,2,1)
se.boxplot(data=df)#findthe outlayers

mp.subplot(2,2,2)
se.histplot(x='total_bill',data=df,kde=True)#frequency destribution.

mp.subplot(2,2,3)
se.barplot(x='total_bill',y='tip',data=df)

```

```python
#correlation
de=df[['total_bill','size','tip']]
de.corr()

#create a plot for correlation. without annot the value are not print on the plot.
se.heatmap(de.corr(),annot=True)
```

```python
#create a pair plot of numerical
se.pairplot(data=df)#see the multiple column correlation.
```

```python
#countplot
#create a count plot of the number of tips by day of the
se.countplot(x='day',data=df)#simila to value_counts()
```

```python
#hue highlight the coulumn value.
se.scatterplot(x='total_bill',y='tip',hue='day',data=df)
```

```python
import matplotlib.pyplot as mp
```

```python
dd=se.load_dataset('iris')
dd
```

```python
# Exercise 1: Create a scatter plot of sepal length vs. petal length, colored by species.
se.scatterplot(data=dd,x='sepal_length',y='petal_length',hue='species')
```

```python
#Calculate the average sepal width for each species. Find the correlation between data.
dd.groupby('species')['sepal_width'].mean()
dd.corr(numeric_only=True)
```

```python
#Find the minimum, maximum, and average sepal length for each species.
dd.groupby('species')['sepal_length'].mean()
```

```python
dd.groupby('species')['sepal_length'].max()
```

```python

dd.groupby('species')['sepal_length'].min()
```

```python
#Create a subplot with a histogram of sepal length and a histogram of petal length.
mp.subplot(2,2,1)
dd.hist('sepal_length')

mp.subplot(2,2,2)
dd.hist('petal_length')
```

```python
#identify the two numerical features with the strongest positive correlation.
dd.corr(numeric_only=True)
se.heatmap(dd.corr(numeric_only=True),)
```

```python
#Filter flowers with petal length greater than 5 and visualize them by species.
filter=dd[dd['petal_length']>5]
filter
se.scatterplot(data=filter,x='sepal_length',y='petal_length',hue='species')
```
