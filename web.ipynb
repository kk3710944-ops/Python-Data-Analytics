```python
import requests

url = "https://www.flipkart.com/search?q=laptop"#is a query parameter. q is the parameter name, and laptop is the value being searched for.

headers = {
    "User-Agent": "Mozilla/5.0"#Mozilla/5.0 is a historical compatibility token commonly included in browser User-Agent strings. 
}

response = requests.get(url, headers=headers)

print(response.status_code)
```

```python
print(response.text[:1000])
```

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(response.text, "html.parser")
```

```python
#Code	Meaning
#soup	The parsed HTML webpage.
#.title	Finds the HTML <title> tag.
#.text	Extracts the text inside that tag.
#print()	Displays the result.
print(soup.title.text)
```

```python
products = soup.find_all("div", class_="RG5Slk")
prices = soup.find_all("div", class_="oFEPlD")

for p, pr in zip(products, prices):#pair the two part od code.p->product and pr->prices
    print(p.text, "=>", pr.text)
```

```python
import pandas as pd

data = []

for p, pr in zip(products, prices):
    data.append([p.text, pr.text])

df = pd.DataFrame(data,
                  columns=["Product", "Price"])
```
