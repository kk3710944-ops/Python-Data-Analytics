```python
import numpy as np
```

```python
#zeros
a=np.zeros((4,4))
print(a)
```

```python
#ones
a=np.ones((4,4))
print(a)
```

```python
#print two
a=np.ones((4,4))
print(a+1)
```

```python
#use range funtion in numpy
a=np.arange(0,10,2)
print(a)
```

```python
#shape our single line of array to matrix format use reshape.[1 2 3 4 5 6 7 8 9]
a=np.arange(1,17).reshape(4,4)
print(a)
#1,5
#2,4
#9,10,13,14
#10,12
#reverse row wise and column wise
print(a[0:2,0])
print(a[0,1:4:2])
print(a[2:4,0:2])
print(a[2,1:4:2])
print(a[::-1])
print(a[0:,::-1])
b=np.arange(1,17).resize(1,5)

print(b)

```
