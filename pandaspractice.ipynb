```python
#Find patients whose BMI > 30 and select only:Glucose,BMI,diabetes Change the diapadies to 1 for patients whose Glucose > 200.
df.loc[(df['bmi'] > 30) & (df['blood_glucose_level'] > 200),'diabetes'] = 1
df.loc[(df['bmi'] > 30)&(df['blood_glucose_level'] > 200),['blood_glucose_level', 'bmi', 'diabetes']]
```

```python
#Find the average Glucose level for each Outcome.
df.groupby('diabetes')['blood_glucose_level'].mean()
```

```python
#Find the average BMI for each Outcome.
df.groupby('diabetes')['bmi'].mean()
```

```python
#Find the number of patients in each Outcome group.
df.groupby('diabetes')['diabetes'].count()
```

```python
#Find the average Glucose for patients with Outcome = 1 and Outcome = 0 separately using groupby().
df.groupby('diabetes')['blood_glucose_level'].mean()
```

```python
#Find the top 5 patients with the highest Glucose using loc/iloc.
df.sort_values('blood_glucose_level',ascending=False).loc[:5]#loc
df.sort_values('blood_glucose_level',ascending=False).iloc[::].head(5)#iloc

```

```python
#Create these age groups:< 30 → Young 30–50 → Middle > 50 → Senior Then find the average Glucose for each age group.
df['age_group']=pd.cut(df['age'],bins=[0,30,50,90],labels=['younger','old','senior'])
df.groupby('age_group')['blood_glucose_level'].mean()
df
```

```python
#Using groupby(), find the average BMI of diabetic patients (Outcome=1) by age group.
de=df[df['diabetes']==1]
de.groupby('age_group')['bmi'].mean()
de.groupby('age_group')['blood_glucose_level'].mean()

```

```python
#Find which Outcome group has the highest average Glucose, BMI, and Age.
df.groupby('diabetes')[['blood_glucose_level','bmi','age']].mean()
```

```python
#create data set
df1=pd.DataFrame({'ID':[1,2,3,4],'NAME':['ram','raj','tom','dom'],'AGE':[21,23,20,21]})
df1
```

```python
df2=pd.DataFrame({'ID':[4,5,6,7],'salary':[50000,40000,28000,55000]})
df2
```

```python
#merge data set
merged_df=pd.merge(df1,df2, on='ID',how='inner')
merged_df
```

```python
#concat
concat_df=pd.concat([df1,df2])
concat_df
```

```python
#pivort_table   tablename|numerical column name|any column name|aggregation function.
p=pd.pivot_table(data=df1,values='ID',index='NAME',columns='AGE',aggfunc='count')
p
```
