```python
f=['apple','banana','orange','mango','graps']
#modify
f[1]='pineapple'
print(f)
#['apple', 'pineapple', 'orange', 'mango', 'graps']

#add
f.append('banana')
print(f)
#['apple', 'pineapple', 'orange', 'mango', 'graps', 'banana']

#remove or delete here we remove the value based on the index value
f.pop(3)
print(f)
#['apple', 'pineapple', 'orange', 'graps', 'banana']

#slicing
a=f[1:3]
print(a)
#['pineapple', 'orange']

#append that add a list on the previous list [1, 2, 3, 4, [5, 6]]
a=[1,2,3,4]
a.append([5,6])
print(a)
#[1, 2, 3, 4, [5, 6]]

#extend that add a elements on the previous list [1, 2, 3, 4, [5, 6], 7, 8]
a.extend([7,8])
print(a)
#[1, 2, 3, 4, [5, 6], 7, 8]

a.extend('ab')
print(a)
#[1, 2, 3, 4, [5, 6], 7, 8, 'a', 'b']

#insert the value based on the index value 
a.insert(3,13)
print(a)
#[1, 2, 3, 13, 4, [5, 6], 7, 8, 'a', 'b']

#remove here we give the actual value but pop we give the index value
a.remove(8)
print(a)
#[1, 2, 3, 13, 4, [5, 6], 7, 'a', 'b']
```
