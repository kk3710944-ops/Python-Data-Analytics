```python
#variable length argument
#sometimes you want to pass a number or argument use kwargs function.that is defined by *.
def s(*a):
    c=sum(a)
    print(c)
s(1,2,3,4)
```

```python
#if you want to pass a value with key use kwargs with **.
def info(**k):
    for k,v in k.items():
        print(f"{k}:{v}")
info(name='ram',age=25,mark=56)
```

```python
#anonymous function
square=lambda x,y:x**y
square(3,5)
```

```python
#default argument
def g(name,age=30):
    print(f"hello {name}!you are {age} years old")
g("ram")
```

```python
#keywords argument
def sub(a,b):
    s=a-b
    return s
sub(b=20,a=25)
```

```python
#positional argument
def g(name,age):
    print(f"hello {name}!you are {age} years old")
g("ram",21)
```
